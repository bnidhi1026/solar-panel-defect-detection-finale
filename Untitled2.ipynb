```python
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt

DATASET_ROOT = Path("/content/drive/MyDrive/Thermal Solar PV Anomaly Detection Dataset")

class_names = [
    'MultiByPassed',
    'MultiDiode',
    'MultiHotSpot',
    'SingleByPassed',
    'SingleDiode',
    'SingleHotSpot',
    'StringOpenCircuit',
    'StringReversedPolarity'
]

IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

def get_image_files(folder):
    if not folder.exists():
        return []
    return sorted([p for p in folder.iterdir() if p.suffix.lower() in IMG_EXTS])

def read_yolo_label(label_path):
    boxes = []
    if not label_path.exists():
        return boxes
    with open(label_path, "r") as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
    for line in lines:
        parts = line.split()
        if len(parts) >= 5:
            cls_id = int(float(parts[0]))
            boxes.append(cls_id)
    return boxes

class_counter = Counter()

for split in ["train", "valid", "test"]:
    img_dir = DATASET_ROOT / "ImageSet" / split / "images"
    lbl_dir = DATASET_ROOT / "ImageSet" / split / "labels"

    image_files = get_image_files(img_dir)
    print(f"\nSplit: {split}, Images found: {len(image_files)}")

    found_labels = 0
    found_boxes = 0

    for img_path in image_files:
        label_path = lbl_dir / f"{img_path.stem}.txt"
        if not label_path.exists():
            continue

        found_labels += 1
        cls_ids = read_yolo_label(label_path)

        for cls_id in cls_ids:
            class_counter[cls_id] += 1
            found_boxes += 1

    print("Matching labels:", found_labels)
    print("Boxes found:", found_boxes)

print("\nClass Counter:", class_counter)

counts = [class_counter.get(i, 0) for i in range(len(class_names))]
print("Counts:", counts)

plt.figure(figsize=(12, 5))
plt.bar(class_names, counts)
plt.xticks(rotation=45, ha='right')
plt.title("Class Distribution")
plt.ylabel("Number of Instances")
plt.tight_layout()
plt.show()
```

```python
# ==============================
# CELL 6: Utility functions
# ==============================

IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

def get_image_files(folder):
    folder = Path(folder)
    return sorted([p for p in folder.iterdir() if p.suffix.lower() in IMG_EXTS])

def get_label_path_from_image(image_path):
    image_path = Path(image_path)
    split_dir = image_path.parent.parent
    labels_dir = split_dir / "labels"
    return labels_dir / (image_path.stem + ".txt")

def read_yolo_label(label_path):
    boxes = []
    if not Path(label_path).exists():
        return boxes
    with open(label_path, "r") as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
    for line in lines:
        parts = line.split()
        if len(parts) >= 5:
            cls_id = int(float(parts[0]))
            x, y, w, h = map(float, parts[1:5])
            boxes.append((cls_id, x, y, w, h))
    return boxes

def yolo_to_xyxy(box, img_w, img_h):
    cls_id, x, y, w, h = box
    x1 = int((x - w/2) * img_w)
    y1 = int((y - h/2) * img_h)
    x2 = int((x + w/2) * img_w)
    y2 = int((y + h/2) * img_h)
    return cls_id, x1, y1, x2, y2
```

```python
# ==============================
# CELL 9: Show sample annotated images
# ==============================

import random
import matplotlib.pyplot as plt
import cv2

def show_random_samples(split="train", n=6):
    img_dir = DATASET_ROOT / "ImageSet" / split / "images"
    image_files = get_image_files(img_dir)

    if len(image_files) == 0:
        print(f"No images found in: {img_dir}")
        return

    sample_files = random.sample(image_files, min(n, len(image_files)))

    rows = (len(sample_files) + 1) // 2
    plt.figure(figsize=(18, 6 * rows))

    for i, img_path in enumerate(sample_files, 1):
        img = cv2.imread(str(img_path))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        h, w = img.shape[:2]

        label_path = DATASET_ROOT / "ImageSet" / split / "labels" / f"{img_path.stem}.txt"
        boxes = read_yolo_label(label_path)

        for box in boxes:
            cls_id, x1, y1, x2, y2 = yolo_to_xyxy(box, w, h)
            cv2.rectangle(img, (x1, y1), (x2, y2), (255, 0, 0), 2)
            cv2.putText(
                img,
                class_names[cls_id],
                (x1, max(20, y1 - 5)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.55,
                (255, 0, 0),
                2
            )

        plt.subplot(rows, 2, i)
        plt.imshow(img)
        plt.axis("off")
        plt.title(img_path.name)

    plt.tight_layout()
    plt.show()

show_random_samples(split="train", n=6)
```

```python
# ==============================
# CELL 10: Resolution and brightness analysis
# ==============================

image_shapes = []
train_img_dir_for_shapes = DATASET_ROOT / "ImageSet" / "train" / "images"
all_train_imgs_for_shapes = get_image_files(train_img_dir_for_shapes)

if len(all_train_imgs_for_shapes) == 0:
    print(f"No training images found in {train_img_dir_for_shapes} for resolution analysis.")
else:
    # Sample up to 500 images for shape analysis to avoid long processing times for very large datasets
    sample_size_for_shapes = min(len(all_train_imgs_for_shapes), 500)
    sampled_imgs_for_shapes = random.sample(all_train_imgs_for_shapes, sample_size_for_shapes)

    for img_path in sampled_imgs_for_shapes:
        img = cv2.imread(str(img_path))
        if img is not None:
            h, w, _ = img.shape
            image_shapes.append((w, h))

widths = [w for w, h in image_shapes]
heights = [h for w, h in image_shapes]

plt.figure(figsize=(6, 4))
plt.hist(widths, bins=20)
plt.title("Image Width Distribution")
plt.xlabel("Width")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(6, 4))
plt.hist(heights, bins=20)
plt.title("Image Height Distribution")
plt.xlabel("Height")
plt.ylabel("Count")
plt.show()

# Brightness estimate
brightness_values = []
train_imgs = get_image_files(DATASET_ROOT / "ImageSet" / "train" / "images")

if len(train_imgs) == 0:
    print("No training images found for brightness analysis.")
else:
    for img_path in random.sample(train_imgs, min(200, len(train_imgs))):
        img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        if img is not None:
            brightness_values.append(img.mean())

    plt.figure(figsize=(6, 4))
    plt.hist(brightness_values, bins=20)
    plt.title("Approximate Thermal Intensity / Brightness Distribution")
    plt.xlabel("Mean Pixel Intensity")
    plt.ylabel("Count")
    plt.show()

print("Brightness mean:", np.mean(brightness_values) if brightness_values else "N/A")
print("Brightness std :", np.std(brightness_values) if brightness_values else "N/A")
```

```python
# ==============================
# CELL 11: Label sanity checks
# ==============================

bad_boxes = []

splits = ["train", "valid", "test"]
for split in splits:
    lbl_dir = DATASET_ROOT / "ImageSet" / split / "labels"
    if not lbl_dir.exists():
        continue

    for label_file in lbl_dir.glob("*.txt"):
        boxes = read_yolo_label(label_file)
        for idx, box in enumerate(boxes):
            cls_id, x, y, w, h = box

            if cls_id < 0 or cls_id >= len(class_names):
                bad_boxes.append((str(label_file), idx, "invalid_class_id", box))

            if not (0 <= x <= 1 and 0 <= y <= 1 and 0 < w <= 1 and 0 < h <= 1):
                bad_boxes.append((str(label_file), idx, "invalid_bbox_range", box))

print("Number of suspicious boxes:", len(bad_boxes))
for item in bad_boxes[:20]:
    print(item)
```

```python
# ==============================
# CELL 12: Load YOLO26m model
# ==============================

model = YOLO("yolo26m.pt")
print("Model loaded successfully.")
```

```python
# ==============================
# CELL 12.5: Create formatted data.yaml for training
# ==============================

FORMATTED_YAML_PATH = DATASET_ROOT / "data_formatted.yaml"

data_config = {
    'path': str(DATASET_ROOT / "ImageSet"),
    'train': 'train/images',
    'val': 'valid/images',
    'test': 'test/images',
    'nc': len(class_names),
    'names': class_names
}

with open(FORMATTED_YAML_PATH, 'w') as f:
    yaml.dump(data_config, f, indent=4)

print("Formatted data.yaml created at:", FORMATTED_YAML_PATH)


# ==============================
# CELL 13: Train the model
# ==============================

results = model.train(
    data=str(FORMATTED_YAML_PATH),
    epochs=100,
    imgsz=640,
    batch=16,              # good starting point for T4; reduce to 8 if OOM
    device=0,
    workers=2,
    project=str(DATASET_ROOT / "runs"),
    name="thermal_pv_yolo26m",
    pretrained=True,
    optimizer="auto",
    cache=False,
    amp=True,
    patience=25,
    cos_lr=True,
    close_mosaic=10,
    save=True,
    verbose=True
)
```

```python
# ==============================
# CELL 14: Validate on validation set
# ==============================

metrics = model.val(
    data=str(FORMATTED_YAML_PATH),
    split="val",
    imgsz=640,
    batch=16,
    device=0
)

print(metrics)
```

```python
# ==============================
# CELL 15: Evaluate on test set
# ==============================

test_metrics = model.val(
    data=str(FORMATTED_YAML_PATH),
    split="test",
    imgsz=640,
    batch=16,
    device=0
)

print(test_metrics)
```

```python
# ==============================
# CELL 16: Inference on test images
# ==============================

best_model_path = DATASET_ROOT / "runs" / "thermal_pv_yolo26m" / "weights" / "best.pt"
print("Best model path:", best_model_path)

trained_model = YOLO(str(best_model_path))

test_images_dir = DATASET_ROOT / "test/images"
sample_test_images = get_image_files(test_images_dir)[:10]

pred_results = trained_model.predict(
    source=[str(p) for p in sample_test_images],
    imgsz=640,
    conf=0.25,
    save=True,
    project=str(DATASET_ROOT / "runs"),
    name="thermal_pv_predictions"
)
```
